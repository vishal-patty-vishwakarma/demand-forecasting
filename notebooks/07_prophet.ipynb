## Notebook 07: Prophet with External Regressors

### What Prophet does differently from SARIMA
- Handles missing dates automatically
- Built-in yearly + weekly seasonality
- Accepts external regressors (oil price, promotions)
- Decomposes forecast into interpretable components

### What we do
1. Run Prophet on same 3 series as SARIMA
2. Add oil_price and discount_depth as regressors
3. Add Ecuador holidays
4. Plot components — trend, weekly, yearly, holiday effects
5. Compare SMAPE to SARIMA and baseline

### Key limitation we document
Prophet is still a per-series model.
Cannot scale to 34,656 series either.
But it shows us what external regressors contribute.
